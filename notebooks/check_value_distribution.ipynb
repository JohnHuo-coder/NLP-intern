```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import json

numeric_df = pd.read_csv("../data/processed/numeric_values.csv")
```

```python
numeric_df.describe()
```

```python
# drop extreme bedroom and bathroom numbers
def remove_outliers(df, col, lower=0.01, upper=0.99):
    low = df[col].quantile(lower)
    high = df[col].quantile(upper)
    
    return df[(df[col] >= low) & (df[col] <= high)]

df_clean = numeric_df.copy()
for col in ['bedrooms', 'bathrooms','price', 'sqft']:
    df_clean = remove_outliers(df_clean, col)
```

```python
df_clean.describe()
# still exists abnormal sqft value, manually set a range
```

```python
df_clean = df_clean[
    (df_clean['sqft'] > 200) & 
    (df_clean['sqft'] < 20000)
]
```

```python
df_clean.describe()
```

```python
cols = ['bedrooms', 'bathrooms', 'price', 'sqft']

stats = {}

for col in cols:
    stats[f"{col}_min"] = float(df_clean[col].min())
    stats[f"{col}_max"] = float(df_clean[col].max())

# 存成 json 文件
with open('../data/processed/stats.json', 'w') as f:
    json.dump(stats, f, indent=2)
```
